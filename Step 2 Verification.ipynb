```python
import pandas as pd
import numpy as np
```

```python
import openpyxl
from openpyxl.styles import Color
```

```python
comparison_df = pd.read_csv("inputFiles/KALRO Test protocol.csv")    
```

```python
df = pd.read_excel("inputFiles/batch-approval-batch_9507 (14).xlsx",sheet_name="Scoring Results",header=1)
```

```python
sample_codes = df['Sample Code'].tolist()
```

```python
sample_codes
```

```python
comparison_df = comparison_df.loc[comparison_df['barcode'].isin(sample_codes)]
```

```python
comparison_df
```

```python
file_path = "inputFiles/batch-approval-batch_9507 (11) (1).xlsx"
sheet_name = "Scoring Results"
columns = ['pH', 'Available P', 'Exchangeable K', 'Organic Matter']  # Columns to check
```

```python
workbook = openpyxl.load_workbook(file_path)
sheet = workbook[sheet_name]
```

```python
sheet.cell(row=15, column=16).fill.start_color.index
```

```python
sheet.cell(row=2, column=15).value
```

```python
codes = {}

codes['FF0000FF'] = 'high'
codes['FF00FF00'] = 'optimum'
codes['FFFFFF00'] = 'low'
codes['FFFF0000'] = 'very low'
```

```python
cell_colors = {}

for col in [15,16,17,35]:
    col_colors = []
    cell_title = sheet.cell(row=2, column=col).value
    cell_colors[cell_title] = []
    print(cell_title)
    for row in range(3, sheet.max_row + 1):
        color = sheet.cell(row=row, column=col).fill.start_color.index
        cell_colors[cell_title].append(codes[color])
```

```python
cell_colors
```

```python
batch = pd.DataFrame(cell_colors, index=comparison_df.barcode)
```

```python
batch = batch.rename(columns={"pH":"ph", "Available P":"phosphorus","Exchangeable K":"potassium"})
```

```python
batch
```

```python
comparison_df = comparison_df.set_index("barcode")
```

```python
comparison_df[batch.columns]
comparison_df = comparison_df.reindex(batch.index)[batch.columns]
```

```python
comparison_df.index == batch.index
```

```python
are_equal = (batch['ph'] == comparison_df['ph']).any()
```

```python
are_equal
```

```python
chemical_class = {}

for column in batch.columns:
    chemical_class[column] = []
    for expected,actual in zip(batch[column],comparison_df[column]):
        if expected == actual:
            chemical_class[column].append(True)
        if expected != actual:
            chemical_class[column].append(False)
```

```python
chemical_class
```

```python
class_comparison_df = pd.DataFrame(chemical_class, index=comparison_df.index)
```
