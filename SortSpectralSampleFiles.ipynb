```python
import pandas as pd
import os
from pathlib import Path
import json
import shutil
```

```python
count = 0
for file in os.listdir("inputFiles/FMAFS_9 batches"):
    print(file)
    os.makedirs(f"outputFiles/mirriam_test/{count}",exist_ok=True)
    df = pd.read_excel(f"inputFiles/FMAFS_9 batches/{file}",sheet_name="Sheet1",engine="openpyxl")
    for index,row in df.iterrows():
        barcode = row['Barcode']
        print(barcode)
        for opus_files in Path(f"./outputFiles/opus_renamed").rglob(f"*{barcode}*"):
            print(opus_files)
            shutil.copyfile(opus_files , f"outputFiles/mirriam_test/{count}/{opus_files.name}")
    count+=1
```
