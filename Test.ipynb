```python
import pandas as pd

df = pd.DataFrame({
    'A': [1, 2, 3, 4, 5],
    'B': [10, 20, 30, 40, 50]
})

def process_columns(df):
    return pd.Series({
        'A_squared': df['A']**2,
        'B_doubled': df['B']*2
    })

df[['A_squared', 'B_doubled']] = df.apply(process_columns, axis=1)

print(df)
```
